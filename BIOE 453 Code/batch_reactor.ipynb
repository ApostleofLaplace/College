### Imports 

```python
from scipy.integrate import odeint
import numpy as np
import matplotlib.pyplot as plt
```

### Function Definitions

```python
def coupled_diff_equ(y, t, mumax, Ks, Yxs):

    X, S = y
    mu = mumax * S / (Ks + S)
    dXdt = mu * X
    dSdt = -dXdt / Yxs

    return [dXdt, dSdt]
```

### Variable Definitions

```python
t = np.linspace(0, 24, 1000)
mumax = 0.35
Ks = 1
Yxs = 0.5
y0 = [0.1, 50.0]
vol = 2
```

### Calling the Functions to make Big Variables

```python
solutions = odeint(coupled_diff_equ, y0, t, args=(mumax, Ks, Yxs))
x_values = solutions[:, 0]
s_values = solutions[:, 1]
p_values = (x_values * vol)
```

### Plot Stuff

```python
plt.plot(t, x_values, label='Biomass (X)')
plt.plot(t, s_values, label='Substrate (S)')
plt.plot(t, p_values, label='Product (P)')
plt.xlabel('time (hr)')
plt.ylabel('concentration (g/L)')
plt.legend()
plt.title('Batch Reactor Plot')
plt.show()
```
