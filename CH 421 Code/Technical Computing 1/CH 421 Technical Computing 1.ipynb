12. Finding the concentration of an unknown

After linear regression, we can also use our calibration curve to determine the concentration of an unknown. First, define an array with the unknown response value(s). 

________________
unknown = np.array([5.6]) % the unknown was measered only 1 time. 
________________

We can calculate the concentration using the line of best fit

___________________
concentration_unknown=(unknown-intercept)/slope 
___________________

We can follow the equations from the textbook or lecture slides to determine the standard uncertainty on the concentrations. It can be helpful to break long equations up into multiple steps 

Recall - we can use len(data) or len(concentration) to get the number of measurements in the calibration curve dataset, n. You will need to define a value for k, the number of repeated measurements of the unknown. 

___________________________

d_i=average_response-(slope*concentration+intercept) # will run calculation for each point in our average response array

s_y=np.sqrt(np.sum(d_i**2)/(len(data)-2)) 

j1=(unknown-np.mean(average_response))**2 

j2=np.sum((concentration-np.mean(concentration))**2)

u_x = (s_y/np.abs(slope))*np.sqrt((1/k)+(1/len(data))+(j1/(j2*(slope**2))))
_______________

Print the unknown concentration and calculated uncertainty. 

```python
# note that this section does not supply a response for the unknown,
# so the example value from the instructions above is used: a single
# measurement of 5.6 a.u. Replace the number in the array (and k) to use a real
# measurement.
unknown = np.array([5.6])   # the unknown was measured only 1 time
k = 1                       # number of repeated measurements of the unknown
n = len(data)               # number of points on the calibration curve, DOF = n - 2

# Two things to be careful about in this equation: ybar is the mean of all the
# calibration responses (not of the unknown), and n is the number of calibration
# points, not the number of replicates.

concentration_unknown = (unknown - intercept)/slope

# standard uncertainty on that concentration, built up in steps
d_i = average_response - (slope*concentration + intercept)  # residual at each calibration point
s_y = np.sqrt(np.sum(d_i**2)/(n - 2))                       # standard deviation of the y residuals

j1 = (unknown - np.mean(average_response))**2
j2 = np.sum((concentration - np.mean(concentration))**2)

u_x = (s_y/np.abs(slope))*np.sqrt((1/k) + (1/n) + (j1/(j2*(slope**2))))

print('unknown response (a.u.)        =', unknown)
print('n (calibration points)         =', n, '   k (replicates of unknown) =', k)
print('s_y (a.u.)                     =', s_y)
print()
print('unknown concentration (mM)     =', concentration_unknown)
print('standard uncertainty, u_x (mM) =', u_x)
print()
print('concentration of the unknown = %.4f +/- %.4f mM' % (concentration_unknown[0], u_x[0]))
```
